# First we need to perform EDA then, modeling

### Now, we need to seperate the our independent feature variables and dependent target variables

```python
y = df['electricity_kwh']
```

```python
X = df[['household_size',
        'rooms',
        'appliances',
        'temp_c',
        'income']]
```

## Step 1: split data into train test split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size =0.3,random_state=42)
```

```python
X_train.shape, X_test.shape
```

# Step 2 : Standard scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

scaler.fit(X_train)
```

```python
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Step 3: Fitting the Model

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_predict = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import r2_score
R_square = r2_score(y_test, y_predict)
print(f"the r2 is : {R_square}")
```
